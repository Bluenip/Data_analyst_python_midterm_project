2. Inspect dataset structure

```python
# Define the paths to the three dataset files
data_paths = {
    "Books": project_path / "archive" / "Books.csv",
    "Users": project_path / "archive" / "Users.csv",
    "Ratings": project_path / "archive" / "Ratings.csv"
}

# Create a dictionary to store the loaded DataFrames
dataframes = {}

# Load and inspect each dataset
for name, path in data_paths.items():

    # Read the CSV file using the encoding commonly used by Book-Crossing
    df = pd.read_csv(path, encoding="latin-1", low_memory=False)

    # Store the DataFrame for later use
    dataframes[name] = df

    # Display the dataset structure
    print("\n" + "=" * 50)
    print(f"Dataset: {name}")
    print(f"Number of rows: {df.shape[0]}")
    print(f"Number of columns: {df.shape[1]}")
    print(f"Column names: {df.columns.tolist()}")

    # Display column data types and non-null counts
    print("\nData information:")
    df.info()
```

```python
# Check data consistency across datasets
for name, df in dataframes.items():

    # Display the dataset name
    print(f"\n{'=' * 50}")
    print(f"Dataset: {name}")

    # Display the data types of all columns
    print("\nData types:")
    print(df.dtypes)

    # Count missing values in each column
    print("\nMissing values:")
    print(df.isnull().sum())

    # Count duplicated rows
    print("\nNumber of duplicate rows:")
    print(df.duplicated().sum())
```

```python
# Check relationships between datasets

# Define the expected relationship columns
relationship_columns = {
    "Books": ["ISBN"],
    "Users": ["User-ID"],
    "Ratings": ["User-ID", "ISBN"]
}

# Check whether each relationship column exists
for table_name, columns in relationship_columns.items():

    # Retrieve the corresponding DataFrame
    df = dataframes[table_name]

    print(f"\nDataset: {table_name}")

    for column in columns:
        # Check whether the expected column exists
        if column in df.columns:
            print(f"{column}: Found")
        else:
            print(f"{column}: NOT FOUND")

print("------------------")
# Check primary key candidates for each dataset
# Define the potential primary keys
primary_keys = {
    "Books": "ISBN",
    "Users": "User-ID"
}

# Check uniqueness and missing values
for table_name, key_column in primary_keys.items():

    # Retrieve the corresponding DataFrame
    df = dataframes[table_name]

    # Count duplicated key values, excluding missing values
    duplicate_count = df[key_column].dropna().duplicated().sum()

    # Count missing key values
    missing_count = df[key_column].isnull().sum()

    print("Primary key check:")
    print(f"Dataset: {table_name}")
    print(f"Key column: {key_column}")
    print(f"Missing key values: {missing_count}")
    print(f"Duplicated key values: {duplicate_count}")

    # Report whether the column satisfies basic primary key requirements
    if missing_count == 0 and duplicate_count == 0:
        print("Result: The column satisfies basic primary key requirements.")
    else:
        print("Result: The column needs further investigation.")

print("------------------")
# Check foreign key relationships between datasets
# Identify ISBN values available in the Books table
book_isbns = set(dataframes["Books"]["ISBN"].dropna())

# Identify user IDs available in the Users table
user_ids = set(dataframes["Users"]["User-ID"].dropna())

# Retrieve the Ratings table
ratings_df = dataframes["Ratings"]

# Find ISBN values in Ratings that do not exist in Books
unknown_books = (
    ~ratings_df["ISBN"].isin(book_isbns)
    & ratings_df["ISBN"].notna()
)

# Find user IDs in Ratings that do not exist in Users
unknown_users = (
    ~ratings_df["User-ID"].isin(user_ids)
    & ratings_df["User-ID"].notna()
)

# Display the results
print("Foreign key integrity check:")
print("Ratings referencing unknown books:", unknown_books.sum())
print("Ratings referencing unknown users:", unknown_users.sum())
print("Missing ISBN values in Ratings:", ratings_df["ISBN"].isnull().sum())
print("Missing User-ID values in Ratings:", ratings_df["User-ID"].isnull().sum())
```

```python
# Estimate CSV file encoding
from charset_normalizer import from_path

# Check the encoding of each dataset file
for table_name, file_path in data_paths.items():

    # Analyze the file and retrieve the best candidate
    result = from_path(file_path).best()

    print(f"\nDataset: {table_name}")

    # Display the estimated encoding if a candidate is available
    if result is not None:
        print("Estimated encoding:", result.encoding)
        print("Encoding confidence information:", result.percent_chaos)
    else:
        print("Encoding could not be estimated reliably.")

# Verify text display
# Display sample book titles and authors
print("Sample book titles and authors:")

print(
    dataframes["Books"][
        ["Book-Title", "Book-Author", "Publisher"]
    ].head(10).to_string(index=False)
)
```
